```python
import pandas as pd
from pathlib import Path

# Search for the dataset in common locations
possible_paths = [
    Path.cwd() / "data" / "creditcard.csv",
    Path.cwd() / "creditcard.csv",
    Path.home() / "Documents" / "Financial_Fraud_Detection" / "data" / "creditcard.csv"
]

DATA_PATH = next((p for p in possible_paths if p.exists()), None)

if DATA_PATH is None:
    print("Dataset dorakaledhu. Current folder:", Path.cwd())
    print("Jupyter lo dataset file location ni select cheyyali.")
else:
    df = pd.read_csv(DATA_PATH)
    print("Dataset loaded successfully!")
    print("Dataset shape:", df.shape)
    display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

# Save graphs in a folder
PLOTS_DIR = Path.cwd() / "visualizations"
PLOTS_DIR.mkdir(exist_ok=True)

# Graph 1: Class distribution
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Class")
plt.title("Credit Card Fraud vs Non-Fraud Transactions")
plt.xlabel("Class (0 = Non-Fraud, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.savefig(PLOTS_DIR / "class_distribution.png", dpi=150)
plt.show()

# Graph 2: Transaction amount distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Amount", bins=50)
plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(PLOTS_DIR / "amount_distribution.png", dpi=150)
plt.show()

# Graph 3: Correlation heatmap
plt.figure(figsize=(12, 9))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Credit Card Dataset Correlation Heatmap")
plt.tight_layout()
plt.savefig(PLOTS_DIR / "correlation_heatmap.png", dpi=150)
plt.show()

print("SUCCESS: All 3 graphs generated!")
print("Graphs saved in:", PLOTS_DIR.resolve())
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDataset statistics:")
display(df.describe())

print("\nColumn data types:")
display(df.dtypes)
```

```python
from pathlib import Path

# Create output folders
output_dir = Path.cwd() / "week1_submission"
plots_dir = output_dir / "visualizations"

plots_dir.mkdir(parents=True, exist_ok=True)

# Clean dataset: remove duplicate rows
df_clean = df.drop_duplicates().copy()

# Save cleaned CSV
df_clean.to_csv(output_dir / "cleaned_creditcard.csv", index=False)

# Copy the 3 graphs into submission folder
import shutil

for graph in [
    "class_distribution.png",
    "amount_distribution.png",
    "correlation_heatmap.png"
]:
    source = Path.cwd() / "visualizations" / graph
    if source.exists():
        shutil.copy2(source, plots_dir / graph)

print("Week 1 files saved!")
print("Cleaned dataset rows:", len(df_clean))
print("Removed duplicate rows:", len(df) - len(df_clean))
print("Submission folder:", output_dir.resolve())
print("Graphs copied:", len(list(plots_dir.glob("*.png"))))
```
