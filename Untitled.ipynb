```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Find the original dataset
possible_paths = [
    Path.cwd() / "data" / "creditcard.csv",
    Path.cwd() / "creditcard.csv",
    Path.home() / "Documents" / "Financial_Fraud_Detection" / "data" / "creditcard.csv"
]

DATA_PATH = next((p for p in possible_paths if p.exists()), None)

if DATA_PATH is None:
    print("Dataset not found. Current folder:", Path.cwd())
else:
    df = pd.read_csv(DATA_PATH)
    print("Dataset loaded successfully!")
    print("Rows and columns:", df.shape)
    print("Column names:", df.columns.tolist())
    display(df.head())
```

```python
WEEK2_DIR = Path.cwd() / "week2_outputs"
WEEK2_DIR.mkdir(parents=True, exist_ok=True)

print("Week 2 folder ready:", WEEK2_DIR.resolve())
```

```python
print("1. Missing values per column:")
display(df.isnull().sum().sort_values(ascending=False).head(10))

print("2. Total missing values:", df.isnull().sum().sum())
print("3. Duplicate rows:", df.duplicated().sum())

print("4. Class distribution:")
display(df["Class"].value_counts())

print("5. Summary statistics:")
display(df.describe().T)
```

```python
# Check outliers in transaction Amount using IQR
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
]

print("Amount lower bound:", lower_bound)
print("Amount upper bound:", upper_bound)
print("Amount outlier rows:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Graph 1: Transaction amount by class
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Class", y="Amount")
plt.ylim(0, df["Amount"].quantile(0.99))
plt.title("Transaction Amount by Class")
plt.xlabel("Class (0 = Non-Fraud, 1 = Fraud)")
plt.ylabel("Transaction Amount")
plt.tight_layout()
plt.savefig(WEEK2_DIR / "amount_by_class.png", dpi=150)
plt.show()

# Graph 2: Correlation with fraud class
class_corr = df.corr(numeric_only=True)["Class"].drop("Class")
class_corr = class_corr.reindex(
    class_corr.abs().sort_values(ascending=False).index
).head(15)

plt.figure(figsize=(9, 6))
sns.barplot(x=class_corr.values, y=class_corr.index)
plt.title("Top 15 Features Correlated with Fraud Class")
plt.xlabel("Correlation with Class")
plt.ylabel("Feature")
plt.tight_layout()
plt.savefig(WEEK2_DIR / "top_feature_correlations.png", dpi=150)
plt.show()

print("Week 2 EDA graphs saved successfully!")
```

```python
from sklearn.preprocessing import StandardScaler

df_clean = df.drop_duplicates().copy()

scaler = StandardScaler()
df_clean["Amount_scaled"] = scaler.fit_transform(
    df_clean[["Amount"]]
)

print("Original dataset shape:", df.shape)
print("After duplicate removal:", df_clean.shape)
print("Amount scaling completed!")
display(df_clean[["Amount", "Amount_scaled", "Class"]].head())
```

```python
import shutil
import json

# Create a separate folder for Week 2 submission
WEEK2_DIR = Path.cwd() / "week2_outputs"
PLOTS_DIR = WEEK2_DIR / "visualizations"
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
df_clean.to_csv(WEEK2_DIR / "cleaned_creditcard_week2.csv", index=False)

# Copy Week 2 graphs
for name in ["amount_by_class.png", "top_feature_correlations.png"]:
    source = Path.cwd() / "week2_outputs" / name
    if source.exists():
        shutil.copy2(source, PLOTS_DIR / name)

# Save data quality summary
summary = {
    "original_rows": int(len(df)),
    "rows_after_duplicate_removal": int(len(df_clean)),
    "duplicate_rows_removed": int(len(df) - len(df_clean)),
    "missing_values_original": int(df.isnull().sum().sum()),
    "amount_outlier_rows": int(len(outliers)),
    "graphs_saved": len(list(PLOTS_DIR.glob("*.png")))
}

with open(WEEK2_DIR / "week2_summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=4)

print("Week 2 files saved!")
print("Folder:", WEEK2_DIR.resolve())
print("Graphs copied:", summary["graphs_saved"])
print("Cleaned dataset saved.")
```
