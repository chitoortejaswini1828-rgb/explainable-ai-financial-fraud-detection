```python
# Select one sample transaction from the dataset
sample_transaction = df.drop(columns=["Class"]).iloc[[0]].copy()

# Scale Time and Amount using the saved scaler
sample_transaction[["Time", "Amount"]] = feature_scaler.transform(
    sample_transaction[["Time", "Amount"]]
)

# Predict transaction
prediction = rf_model.predict(sample_transaction)[0]
probability = rf_model.predict_proba(sample_transaction)[0][1]

print("Prediction Demo Result")
print("----------------------")

if prediction == 1:
    print("Result: Potential Fraudulent Transaction")
else:
    print("Result: Normal Transaction")

print("Fraud probability:", round(float(probability), 4))
print("Note: This is a sample from the existing dataset.")
```

```python
from pathlib import Path
from xml.sax.saxutils import escape
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, PageBreak
)

project_dir = Path.home() / "Financial_Fraud_Detection_Final"
report_dir = project_dir / "reports"
report_md = report_dir / "Final_Project_Report.md"
report_pdf = report_dir / "Final_Project_Report.pdf"

text = report_md.read_text(encoding="utf-8")

doc = SimpleDocTemplate(
    str(report_pdf),
    pagesize=A4,
    rightMargin=0.7 * inch,
    leftMargin=0.7 * inch,
    topMargin=0.7 * inch,
    bottomMargin=0.7 * inch
)

styles = getSampleStyleSheet()
styles.add(ParagraphStyle(
    name="ReportTitle", parent=styles["Title"],
    alignment=TA_CENTER, fontSize=18, leading=23,
    spaceAfter=20
))
styles.add(ParagraphStyle(
    name="SectionHeading", parent=styles["Heading2"],
    fontSize=13, leading=17, spaceBefore=12, spaceAfter=7
))
styles.add(ParagraphStyle(
    name="ReportBody", parent=styles["BodyText"],
    fontSize=10, leading=15, spaceAfter=6
))

story = []
first_heading = True

for line in text.splitlines():
    line = line.strip()
    if not line:
        story.append(Spacer(1, 4))
    elif line.startswith("# "):
        title = line[2:].strip()
        if first_heading:
            story.append(Paragraph(escape(title), styles["ReportTitle"]))
            first_heading = False
        else:
            story.append(Paragraph(escape(title), styles["SectionHeading"]))
    elif line.startswith("## "):
        story.append(Paragraph(
            escape(line[3:].strip()), styles["SectionHeading"]
        ))
    elif line.startswith("- "):
        story.append(Paragraph(
            "&#8226; " + escape(line[2:]), styles["ReportBody"]
        ))
    else:
        story.append(Paragraph(escape(line), styles["ReportBody"]))

doc.build(story)

print("PDF CREATED:", report_pdf.exists())
print("Saved at:", report_pdf)
print("Size:", round(report_pdf.stat().st_size / 1024, 2), "KB")
```
